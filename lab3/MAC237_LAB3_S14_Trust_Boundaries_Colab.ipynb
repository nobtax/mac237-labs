## Section 9. One row per boundary

Fill every field: one threat that crosses the boundary, the mitigation you would apply, and a
cost of low, medium or high. Then rank the three boundaries in the order you would buy the
mitigations. The check at the bottom refuses an empty field and a cost outside the three
values, and nothing else: it cannot tell a sharp answer from a vague one, so that part is yours.

```python
BOUNDARIES = {
    'browser to web tier': {
        'threat': 'Reflected XSS from unescaped user input',
        'mitigation': 'Context-aware output encoding and input validation',
        'cost': 'low'
    },
    'web tier to application tier': {
        'threat': 'Unauthorized requests reaching application functions',
        'mitigation': 'Server-side authorization and strict request validation',
        'cost': 'medium'
    },
    'application tier to database': {
        'threat': 'SQL injection through unsafe database queries',
        'mitigation': 'Parameterized SQL queries and least-privilege database access',
        'cost': 'medium'
    }
}

RANKING = [
    'application tier to database',
    'web tier to application tier',
    'browser to web tier'
]

print('token:', TOKEN)
blank = [f'{b}.{k}' for b, row in BOUNDARIES.items() for k, v in row.items() if not v.strip()]
assert not blank, 'fill these: ' + ', '.join(blank)
bad_cost = [b for b, row in BOUNDARIES.items() if row['cost'] not in ('low', 'medium', 'high')]
assert not bad_cost, 'cost must be low, medium or high: ' + ', '.join(bad_cost)
assert sorted(RANKING) == sorted(BOUNDARIES), 'RANKING must list each boundary exactly once'
for b in RANKING:
    row = BOUNDARIES[b]
    print(f"{b:30s} {row['cost']:7s} threat: {row['threat']}")
    print(f"{'':30s} {'':7s} mitigation: {row['mitigation']}")
print()
print('  CHECK PASS: three boundaries, each with a threat, a mitigation and a cost, ranked')
```

### Which section attacks which boundary

Five of the sections in this lab attack one of those boundaries. Name them. Each value must be
one of the three boundary names above.

```python
SECTION_MAP = {
    'Section 3, handling program input': 'web tier to application tier',
    'Section 4, memory safety': 'web tier to application tier',
    'Section 5, HTTPS and TLS': 'browser to web tier',
    'Section 7, reflected XSS': 'browser to web tier',
    'Section 8, same-origin policy': 'browser to web tier',
}

print('token:', TOKEN)
assert len(SECTION_MAP) == 5, 'name exactly five sections'
wrong = [s for s, b in SECTION_MAP.items() if b not in BOUNDARIES]
assert not wrong, 'these do not name a boundary: ' + ', '.join(wrong)

# The diagram, drawn from your own table.
tiers = ['browser', 'web tier', 'application tier', 'database']
names = list(BOUNDARIES)
for i, tier in enumerate(tiers):
    print(f'  [ {tier:^18s} ]')
    if i < len(names):
        b = names[i]
        hits = [s.split(',')[0] for s, x in SECTION_MAP.items() if x == b]
        print(f"  ---- trust boundary: {b} ({BOUNDARIES[b]['cost']}) {', '.join(hits)}")
print()
print('  CHECK PASS: five sections mapped to the boundary each one attacks')
```
